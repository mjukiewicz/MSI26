```python
import pandas as pd
import matplotlib.pyplot as plt

# Wczytanie historii
df = pd.read_csv('history.csv')

# Wykresy
plt.figure(figsize=(12, 4))

plt.plot(df['accuracy'], label='Trening')
plt.plot(df['val_accuracy'], label='Test')
plt.title('Wykres zmian wartości Accuracy podczas uczenia')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()

plt.show()
```
