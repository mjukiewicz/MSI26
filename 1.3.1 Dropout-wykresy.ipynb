```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator

# Wczytanie historii
df_with = pd.read_csv('history_with_dropout.csv')
df_without = pd.read_csv('history_without_dropout.csv')

# Numery epok od 1
epochs = range(1, len(df_with) + 1)

# Wykresy
plt.figure(figsize=(12, 4))

ax1=plt.subplot(1, 2, 1)
plt.plot(epochs,df_with['accuracy'], label='Trening')
plt.plot(epochs,df_with['val_accuracy'], label='Test')
plt.title('Model z Dropout')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()
ax1.xaxis.set_major_locator(MaxNLocator(integer=True))

ax2=plt.subplot(1, 2, 2)
plt.plot(epochs,df_without['accuracy'], label='Trening')
plt.plot(epochs,df_without['val_accuracy'], label='Test')
plt.title('Model bez Dropout')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()
ax2.xaxis.set_major_locator(MaxNLocator(integer=True))

plt.tight_layout()
plt.show()
```
